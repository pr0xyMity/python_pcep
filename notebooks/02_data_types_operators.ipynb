## 2.1 Variables

Variables are containers for storing data values.

```python
# Creating variables
name = "Alice"
age = 25
height = 5.6
is_student = True

print("Name:", name)
print("Age:", age)
print("Height:", height)
print("Student:", is_student)
```

### Variable Naming Rules:
- Must start with a letter or underscore
- Can contain letters, numbers, and underscores
- Case-sensitive
- Cannot use Python keywords

```python
# Valid variable names
my_variable = 10
_private = 20
variable2 = 30

# Invalid variable names (commented out)
# 2variable = 40  # Cannot start with a number
# my-variable = 50  # Cannot use hyphens
# for = 60  # Cannot use keywords
```

## 2.2 Data Types

Python has several built-in data types:

```python
# Integer (int)
x = 42
print("Integer:", x, "Type:", type(x))

# Float
y = 3.14
print("Float:", y, "Type:", type(y))

# String (str)
z = "Hello"
print("String:", z, "Type:", type(z))

# Boolean (bool)
w = True
print("Boolean:", w, "Type:", type(w))
```

## 2.3 Type Conversion

Converting between different data types:

```python
# Integer to float
num = 10
float_num = float(num)
print("Float:", float_num)

# Float to integer (truncates decimal part)
decimal = 3.99
int_num = int(decimal)
print("Integer:", int_num)

# Number to string
value = 42
text = str(value)
print("String:", text, "Type:", type(text))

# String to number
text_num = "123"
number = int(text_num)
print("Number:", number)
```

## 2.4 Arithmetic Operators

Python supports standard arithmetic operations:

```python
# Basic operations
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)  # Returns integer part
print("Modulus:", a % b)  # Returns remainder
print("Exponentiation:", a ** b)  # a to the power of b
```

## 2.5 Comparison Operators

Used to compare values, return True or False:

```python
x = 10
y = 5

print("Equal:", x == y)
print("Not equal:", x != y)
print("Greater than:", x > y)
print("Less than:", x < y)
print("Greater or equal:", x >= y)
print("Less or equal:", x <= y)
```

## 2.6 Logical Operators

Used to combine conditional statements:

```python
a = True
b = False

print("AND:", a and b)  # True if both are True
print("OR:", a or b)   # True if at least one is True
print("NOT:", not a)   # Inverts the boolean value
```
