```python
# First understanding what a neural network is
# What are the basic building blocks of a neural network

# Lets start with the structure
# A neural network is made up of input layers, hidden layers, and output layers
# Each layer is made up of nodes, which are also called neurons

# Already we are talking about layers and nodes but what are they?

input = [1, 2, 3]
weights = [0.2, 0.8, -0.5]
bias = 2

# Now lets calculate the output of this simple neural network
output = input[0] * weights[0] + input[1] * weights[1] + input[2] * weights[2] + bias
print(output)
```

```python
# Now that we have the basic input that takes 3 values, and one neuron that produces one output based on the input values and weights and bias, we can start to build a more complex scenario

inputs = [1, 2, 3, 2.5]

weights = [
		[0.2, 0.8, -0.5, 1.0],
		[0.5, -0.91, 0.26, -0.5],
		[-0.26, -0.27, 0.17, 0.87]
]

bias1 = 2
bias2 = 3
bias3 = 0.5

output = [
		inputs[0] * weights[0][0] + inputs[1] * weights[0][1] + inputs[2] * weights[0][2] + inputs[3] * weights[0][3] + bias1,
		inputs[0] * weights[1][0] + inputs[1] * weights[1][1] + inputs[2] * weights[1][2] + inputs[3] * weights[1][3] + bias2,
		inputs[0] * weights[2][0] + inputs[1] * weights[2][1] + inputs[2] * weights[2][2] + inputs[3] * weights[2][3] + bias3	
]

print(output)

```

```python
# The first excercise is very static we can do better by using loops
inputs = [1, 2, 3, 2.5]

weights = [
		[0.2, 0.8, -0.5, 1.0],
		[0.5, -0.91, 0.26, -0.5],
		[-0.26, -0.27, 0.17, 0.87]
]

biases = [2, 3, 0.5]

layer_outputs = []

print(zip(weights, biases))

for neuron_weights, neuron_bias in zip(weights, biases):
		neuron_output = 0
		for n_input, weight in zip(inputs, neuron_weights):
				neuron_output += n_input * weight
		neuron_output += neuron_bias
		layer_outputs.append(neuron_output)
```
