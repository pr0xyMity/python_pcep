## 1.1 What is Python?

Python is a high-level, interpreted, general-purpose programming language.

### Key Characteristics:
- **Easy to learn**: Clear syntax, readable code
- **Interpreted**: No compilation needed
- **High-level**: Abstract from hardware details
- **General-purpose**: Can be used for many types of applications

## 1.2 Your First Python Program

The traditional first program in any language:

```python
# This is a comment
print("Hello, World!")
```

### Understanding the Code:
- `print()` is a built-in function
- The text inside quotes is a string
- Comments start with `#` and are ignored by Python

## 1.3 Python Literals

Literals are data whose values are determined by the literal itself.

```python
# Integer literals
print(123)
print(-456)

# Floating-point literals
print(3.14)
print(-0.5)

# String literals
print("Hello")
print('World')
```

## 1.4 Basic Output with print()

The `print()` function displays output to the console.

```python
# Printing multiple items
print("Python", "is", "awesome")

# Printing with different separators
print("Python", "is", "awesome", sep="-")

# Printing without newline
print("Hello", end=" ")
print("World")
```
