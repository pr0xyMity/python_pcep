## 4.1 Functions

Functions are reusable blocks of code:

```python
# Simple function
def greet():
    print("Hello, World!")

# Call the function
greet()
```

```python
# Function with parameters
def greet_person(name):
    print(f"Hello, {name}!")

greet_person("Alice")
greet_person("Bob")
```

```python
# Function with multiple parameters
def add_numbers(a, b):
    print(f"Sum of {a} and {b} is {a + b}")

add_numbers(5, 3)
add_numbers(10, 20)
```

## 4.2 Return Values

Functions can return values:

```python
# Function with return value
def multiply(a, b):
    return a * b

result = multiply(4, 5)
print("Result:", result)

# Using return value in expression
print("Double result:", multiply(4, 5) * 2)
```

```python
# Function with multiple return values
def get_name_and_age():
    name = "Alice"
    age = 25
    return name, age

person_name, person_age = get_name_and_age()
print(f"Name: {person_name}, Age: {person_age}")
```

## 4.3 Default Parameters

Parameters can have default values:

```python
def greet_with_title(name, title="Mr."):
    print(f"Hello, {title} {name}!")

greet_with_title("Smith")  # Uses default title
greet_with_title("Smith", "Dr.")  # Custom title
```

## 4.4 Tuples

Tuples are immutable sequences:

```python
# Creating tuples
coordinates = (10, 20)
rgb_color = (255, 128, 0)
person = ("Alice", 25, "Engineer")

print("Coordinates:", coordinates)
print("Color:", rgb_color)
print("Person:", person)
```

```python
# Accessing tuple elements
point = (5, 10, 15)

print("X:", point[0])
print("Y:", point[1])
print("Z:", point[2])

# Tuple unpacking
x, y, z = point
print(f"Unpacked: x={x}, y={y}, z={z}")
```

```python
# Tuples are immutable (cannot be changed)
numbers = (1, 2, 3)
# numbers[0] = 10  # This would cause an error!

# But you can create a new tuple
new_numbers = (10,) + numbers[1:]
print("New tuple:", new_numbers)
```

## 4.5 Dictionaries

Dictionaries store key-value pairs:

```python
# Creating dictionaries
student = {
    "name": "Alice",
    "age": 20,
    "grade": "A"
}

print("Student:", student)
print("Name:", student["name"])
print("Age:", student["age"])
```

```python
# Dictionary operations
person = {"name": "Bob", "age": 30}

# Add new key-value pair
person["city"] = "New York"
print("After adding city:", person)

# Update existing value
person["age"] = 31
print("After updating age:", person)

# Remove key-value pair
del person["city"]
print("After removing city:", person)
```

```python
# Dictionary methods
car = {"brand": "Toyota", "model": "Camry", "year": 2020}

# Get all keys
print("Keys:", car.keys())

# Get all values
print("Values:", car.values())

# Get all items (key-value pairs)
print("Items:", car.items())

# Safe get with default
print("Color:", car.get("color", "unknown"))
```

```python
# Iterating through dictionaries
scores = {"Alice": 95, "Bob": 87, "Charlie": 92}

# Iterate over keys
for name in scores:
    print(f"{name}: {scores[name]}")

# Iterate over key-value pairs
for name, score in scores.items():
    print(f"{name} scored {score}")
```

## 4.6 Exception Handling

Handle errors gracefully:

```python
# Basic try-except
try:
    result = 10 / 0  # This will cause an error
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
```

```python
# Handling multiple exceptions
try:
    number = int("abc")  # This will cause ValueError
except ValueError:
    print("Error: Invalid number format!")
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
```

```python
# try-except-else-finally
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Error occurred!")
else:
    print("Success! Result:", result)
finally:
    print("This always executes")
```
