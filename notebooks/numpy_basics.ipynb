```python
# Boolean Indexing, Set Operations, and Sorting in NumPy
import numpy as np

x = np.arange(25).reshape(5, 5)

print("Original Array:")
print(x)	


# We can use boolean indexing to also specify the place where we want to modify

x[(x>10) & (x<20)] = -1

print("\nModified Array:")
print(x)

# Sorting
y = np.random.randint(0, 100, size=(100,))
print("\nUnsorted Array:")
print(y)

y_sorted = np.sort(y)
print("\nSorted Array:")
print(y_sorted)
print("\n Original Array after sorting (remains unchanged):")
print(y)
```

```python
# Simple neural network example using NumPy arrays

x = np.array([[2], [-1], [0.5]]).reshape(3,1)
print("\nInput:\n", x)

weights = np.array([[0.1, 1.0], [-0.3, 0.5], [2.0, -1.0]]).reshape(3,2)
print("\nWeights:\n", weights)

bias = np.array([[0.2], [-0.4]]).reshape(2,1)
print("\nBias:\n", bias)

# First input neuron ftakes 3 weights from Weights matrix and does a scalar product
# I need to do y = x^T * weights + bias
y = np.dot(x.T, weights) + bias.T
print("\nOutput:\n", y)	

# Every single row of the input data is being multiplied by the weight number
# We create a count matrix in order to count in at once using a GPU not doing loops in loopes
```
